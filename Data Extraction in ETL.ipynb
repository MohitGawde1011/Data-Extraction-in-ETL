1. Describe different types of data sources used in ETL with suitable examples.

 -  ETL (Extract, Transform, Load) collects data from different sources. Common data sources are:

Relational Databases (RDBMS)
Stores data in tables.
Examples: MySQL, Oracle, SQL Server, PostgreSQL.


Flat Files
Simple text-based files.
Examples: CSV, TXT.


Excel Files
Spreadsheet files used for reports and analysis.
Examples: .xls, .xlsx.


APIs (Application Programming Interfaces)
Provide data from web applications.
Examples: Weather API, Twitter API, Google Maps API.


Cloud Storage
Stores files and data online.
Examples: AWS S3, Google Cloud Storage, Azure Blob Storage.


NoSQL Databases
Used for unstructured or semi-structured data.
Examples: MongoDB, Cassandra.

2. What is data extraction? Explain its role in the ETL pipeline.

-  Data extraction is the process of collecting data from one or more source systems before transforming and loading it into a target system.

Role in ETL:

Collects data from multiple sources.
Ensures accurate and complete data is extracted.
Supplies raw data for transformation.
Acts as the first step of the ETL process.

Example:
Extract customer records from a MySQL database and sales data from a CSV file before combining them in a data warehouse.

3. Explain the difference between CSV and Excel in terms of extraction and ETL usage.

-  | CSV                     | Excel                                     |
| ----------------------- | ----------------------------------------- |
| Plain text file         | Spreadsheet file                          |
| Faster to read          | Slower due to formatting                  |
| Stores only data        | Stores data, formulas, charts, formatting |
| Smaller file size       | Larger file size                          |
| Easy to automate        | Requires Excel libraries/tools            |
| Best for large ETL jobs | Best for reports and manual editing       |

5. Explain three common challenges faced during data extraction.

-  Poor Data Quality
Missing values, duplicate records, or incorrect data.
Large Data Volume
Extracting huge datasets takes more time and resources.
Different Data Formats
Data may come from databases, CSV files, Excel files, APIs, and XML/JSON, making integration difficult.

6. What are APIs? Explain how APIs help in real-time data extraction.

-  APIs (Application Programming Interfaces) allow applications to communicate and exchange data.

How APIs help in real-time extraction:

Provide live and updated data.
Enable automatic data retrieval.
Reduce manual data collection.
Support JSON and XML data formats.
Allow integration with web applications and cloud services.

Example:
A weather application uses a Weather API to fetch live temperature and forecast data.

7. Why are databases preferred for enterprise-level data extraction?

-  Databases are preferred because they:

Store large volumes of structured data.
Support fast SQL queries.
Maintain data integrity and consistency.
Provide security through authentication and permissions.
Support backup and recovery.
Allow multiple users to access data simultaneously.
Easily integrate with ETL tools.

8. What steps should an ETL developer take when extracting data from large CSV files (1GB+)?
-  Read the file in small chunks instead of loading the entire file into memory.
Remove unnecessary columns during extraction.
Validate and clean the data while reading.
Handle missing or corrupt records.
Use efficient ETL tools or programming libraries.
Monitor memory and CPU usage.
Log extraction progress and errors.
Load processed data into the target database in batches.

Benefits:

Reduces memory usage.
Improves performance.
Prevents system crashes.
Makes ETL processing faster and more reliable.
